# 05. DipDECK library loss: prott5_pca113_full, seeds 0 to 9

**1,010 fits:** ten blocks, 041 to 050.
Each block runs one whole-record reference followed by its 100-fit fold grid.
Only reconstruction loss changes: ClustPy `mean_squared_error`, summing features and averaging
records, instead of the standalone's feature-averaged `MSELoss`. The original factory, fold split,
training-only PCA, fitting, prediction, scores and archive validators are reused.

**Run:** upload this project folder to Google Drive as `MyDrive/explaining-plm-latent-spaces` (or set `PROJECT_DIR`), select **T4 GPU**, then **Run all**.
The small loss adapter is embedded in this notebook.
Setup runs once. The final code cell runs all ten blocks sequentially and stops on any error.

Run the lowest-numbered notebook not already running or completed. Never run this same number in
two runtimes. Completed stages are validated and skipped. An interrupted grid repeats its
100 fits; its completed reference is retained. A stale `.running` file must be removed only after
confirming that block's previous runtime has stopped.

Each reference is validated and archived before its grid. Each complete block is archived before
the next starts, so completed results can be used without waiting for this notebook to finish.
Working outputs go to:
`DRIVE_ROOT/results/dipdeck_library_loss/<block_id>/`.
Reference and complete ZIPs go to `DRIVE_ROOT/dipdeck_library_loss_archives/`.
Each complete ZIP holds one block's 101 fits.
After interruption, run this notebook again.
