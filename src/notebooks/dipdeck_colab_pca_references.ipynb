# DipDECK PCA references, five jobs

Runs the paper-default reference (K=35, merge threshold 0.90, seeds 0 to 9) on all GraphPart-retained records in five PCA spaces: 67 deduplicated, 113 full and deduplicated, 34 full and deduplicated. PCA 67 full is in `dipdeck_colab_pca67_reference.ipynb`. PCA is fitted on all and only that job's records. About 30 to 40 minutes in total.

Setup runs once: the project copy, the patched DipDECK source, the recorded CUDA 12.8 torch build and the version check. Each job cell then fits, validates and writes its own ZIP to `DRIVE_ROOT/`, with the same name the single-job notebooks use. A job whose output on Drive already validates is skipped, so after a disconnect, start a fresh runtime and **Run all** again. Any failure stops the notebook. The last cell prints every archive and its SHA256.

Upload this project folder to Google Drive as `MyDrive/explaining-plm-latent-spaces` (or set `PROJECT_DIR`). Select **T4 GPU**, use a fresh runtime, and run all cells.

```python
import hashlib
import os
import shutil
import subprocess
import sys
from pathlib import Path

standalone = repo / "src/algo/dipdeck/dipdeck_standalone.py"
if not standalone.is_file():
    raise FileNotFoundError(f"Standalone is missing: {standalone}")
output_root = (drive_root / "results/dipdeck").resolve()
standalone_env = os.environ.copy()
standalone_env["DIPDECK_OUTPUT_ROOT"] = str(output_root)
completed_jobs = []

MODES = {
    "reference": (
        "scientific_reference", "--run-scientific-pca-reference",
        "--validate-scientific-pca-reference-output",
    ),
    "grid": ("scientific_grid", "--run-scientific-pca", "--validate-scientific-pca-output"),
}


def run_job(kind: str, pca_dim: int, dedup: bool) -> None:
    prefix, run_flag, validate_flag = MODES[kind]
    owned_directory = output_root / f"{prefix}_pca{pca_dim}{'_dedup' if dedup else ''}"
    command = [
        sys.executable, str(standalone), run_flag, "--pca-dim", str(pca_dim),
        "--device", "cuda", "--clustpy-source-root", str(clustpy_source),
    ]
    validator = [
        sys.executable, str(standalone), validate_flag, "--pca-dim", str(pca_dim),
        "--output-root", str(output_root),
    ]
    if dedup:
        command.append("--dedup")
        validator.append("--dedup")
    archive_base = drive_root / f"dipdeck_{owned_directory.name}_outputs"
    archive_path = archive_base.with_suffix(".zip")
    already_valid = False
    if owned_directory.exists():
        already_valid = subprocess.run(validator, cwd=repo).returncode == 0
        if not already_valid:
            print(f"Replacing incomplete output: {owned_directory}")
    if already_valid:
        print(f"SKIP {owned_directory.name}: output on Drive already validates.")
    else:
        subprocess.run(command, cwd=repo, env=standalone_env, check=True)
        subprocess.run(validator, cwd=repo, check=True)
        archive_path.unlink(missing_ok=True)
    if not archive_path.is_file():
        shutil.make_archive(
            str(archive_base), "zip", root_dir=output_root, base_dir=owned_directory.name,
        )
    digest = hashlib.sha256(archive_path.read_bytes()).hexdigest()
    completed_jobs.append((owned_directory.name, archive_path.name, digest))
    print(f"Output archive: {archive_path}")
    print(f"Output SHA256: {digest}")
```
